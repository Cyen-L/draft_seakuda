```python
import cv2

# 关键修改：显式指定使用 DirectShow 后端
cap = cv2.VideoCapture(1, cv2.CAP_DSHOW) 

if not cap.isOpened():
    print("无法打开摄像头")
    # 此处可以添加一些诊断信息
else:
    print("摄像头已成功打开！")
    # 后续的读取和显示代码...

# 检查摄像头是否成功打开
if not cap.isOpened():
    print("无法打开摄像头")
    exit()

# ---------- 加载模板 ----------
template = cv2.imread('battery_map.jpg', 0)  # 你的参考图片，灰度模式
h, w = template.shape

# 2. 循环读取帧
while True:
    # ret 为布尔值，表示是否成功读取；frame 为读取到的图像帧 (numpy array)
    ret, frame = cap.read()
    
    if not ret:
        print("无法获取画面")
        break

    # 3. 显示画面
    cv2.imshow('USB Camera', frame)

    # 4. 等待按键，按 'ESC' (ASCII码 27) 退出
    key = cv2.waitKey(1)
    if key == 27:  # ESC键
        break

# 5. 释放资源
cap.release()
cv2.destroyAllWindows()
```
